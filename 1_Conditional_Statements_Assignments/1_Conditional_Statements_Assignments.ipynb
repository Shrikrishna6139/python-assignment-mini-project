### Q23
Write a program to calculate electricity bill based on units consumed.

```python
# First 100 units Rs.5, next 100 units Rs.7, above 200 units Rs.10
units = int(input("Enter units consumed: "))

if units < 0:
    print("Units cannot be negative")
else:
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = 100 * 5 + (units - 100) * 7
    else:
        bill = 100 * 5 + 100 * 7 + (units - 200) * 10
    print("Electricity bill: Rs.", bill)
```

### Q25
Write a program to check whether a triangle is equilateral, isosceles, or scalene.

```python
a = float(input("Enter side 1: "))
b = float(input("Enter side 2: "))
c = float(input("Enter side 3: "))

# sides must be positive and satisfy triangle inequality
if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("These sides cannot form a triangle")
elif a == b == c:
    print("Equilateral triangle")
elif a == b or b == c or a == c:
    print("Isosceles triangle")
else:
    print("Scalene triangle")
```

### Q40
Write a program to check whether a password is valid using nested if.

```python
# Rules: length >= 8, contains a digit, contains an uppercase letter
password = input("Enter a password: ")

if len(password) >= 8:
    if any(c.isdigit() for c in password):
        if any(c.isupper() for c in password):
            print("Valid password")
        else:
            print("Invalid: needs at least one uppercase letter")
    else:
        print("Invalid: needs at least one digit")
else:
    print("Invalid: needs at least 8 characters")
```

### Q41
Write a program to print weekday name using match case.

```python
# match-case needs Python 3.10 or above
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number")
```

### Q42
Write a program to perform calculator operations using match case.

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+, -, *, /, %): ")

match op:
    case "+":
        print("Result:", a + b)
    case "-":
        print("Result:", a - b)
    case "*":
        print("Result:", a * b)
    case "/":
        if b == 0:
            print("Cannot divide by zero")
        else:
            print("Result:", a / b)
    case "%":
        if b == 0:
            print("Cannot divide by zero")
        else:
            print("Result:", a % b)
    case _:
        print("Invalid operator")
```

### Q50
Write a program to build a simple ATM menu using a match case.

```python
# Simple ATM menu (no loop, so it handles one choice per run)
balance = 10000
print("---- ATM MENU ----")
print("1. Check balance")
print("2. Deposit")
print("3. Withdraw")
print("4. Exit")
choice = int(input("Enter your choice: "))

match choice:
    case 1:
        print("Current balance: Rs.", balance)
    case 2:
        amount = float(input("Enter deposit amount: "))
        if amount > 0:
            balance += amount
            print("Deposited. New balance: Rs.", balance)
        else:
            print("Invalid amount")
    case 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0:
            print("Invalid amount")
        elif amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Please collect your cash. Balance: Rs.", balance)
    case 4:
        print("Thank you for using our ATM")
    case _:
        print("Invalid choice")
```
