### Q18

### Q19
Create a list of 10 numbers and remove duplicates.

```python
nums = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
unique = []

for x in nums:
    if x not in unique:
        unique.append(x)

print("Original:", nums)
print("Without duplicates:", unique)
```

### Q33
Create a list and find the second largest number.

```python
nums = [12, 45, 2, 41, 45, 31]
unique = sorted(set(nums))

if len(unique) >= 2:
    print("Second largest:", unique[-2])
else:
    print("Not enough distinct elements")
```

### Q34
Create a list and find common elements between two lists.

```python
a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
common = []

for x in a:
    if x in b and x not in common:
        common.append(x)

print("Common elements:", common)
```

### Q35

### Q36

### Q37

### Q39
Create a nested list and access inner elements.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Whole list:", matrix)
print("First inner list:", matrix[0])
print("Element at row 1, column 2:", matrix[1][2])
print("Last element:", matrix[-1][-1])
```

### Q40
Create a list and demonstrate all list methods.

```python
nums = [5, 3, 8]
print("Start:", nums)

nums.append(10);        print("append(10)     ->", nums)
nums.insert(1, 99);     print("insert(1, 99)  ->", nums)
nums.extend([7, 3]);    print("extend([7,3])  ->", nums)
nums.remove(99);        print("remove(99)     ->", nums)
print("pop()          ->", nums.pop(), nums)
print("index(8)       ->", nums.index(8))
print("count(3)       ->", nums.count(3))
nums.sort();            print("sort()         ->", nums)
nums.reverse();         print("reverse()      ->", nums)
copied = nums.copy();   print("copy()         ->", copied)
nums.clear();           print("clear()        ->", nums)
```
