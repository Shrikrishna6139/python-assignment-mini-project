### Q4
Count the number of vowels in a string.

```python
text = input("Enter a string: ").lower()
count = 0
for v in "aeiou":
    count += text.count(v)
print("Vowels:", count)
```

### Q5
Count the number of consonants in a string.

```python
text = input("Enter a string: ").lower()
count = 0
for ch in text:
    if ch.isalpha() and ch not in "aeiou":
        count += 1
print("Consonants:", count)
```

### Q23

```python
text = input("Enter a string: ").lower().replace(" ", "")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

### Q26
Find the first and last occurrence of a character.

```python
text = input("Enter a string: ")
ch = input("Enter a character: ")

if ch in text:
    print("First occurrence at index", text.find(ch))
    print("Last occurrence at index", text.rfind(ch))
else:
    print("Character not found")
```

### Q31
Remove punctuation marks from a string.

```python
import string

text = input("Enter a string: ")
clean = ""
for ch in text:
    if ch not in string.punctuation:
        clean += ch
print(clean)
```

### Q36
Remove duplicate characters from a string.

```python
text = input("Enter a string: ")
result = ""
for ch in text:
    if ch not in result:
        result += ch
print("Without duplicates:", result)
```

### Q37
Find the longest word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
longest = words[0]
for w in words:
    if len(w) > len(longest):
        longest = w
print("Longest word:", longest)
```

### Q38
Find the shortest word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
shortest = words[0]
for w in words:
    if len(w) < len(shortest):
        shortest = w
print("Shortest word:", shortest)
```

### Q40
Count uppercase and lowercase letters separately.

```python
text = input("Enter a string: ")
upper = 0
lower = 0
for ch in text:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
print("Uppercase letters:", upper)
print("Lowercase letters:", lower)
```
